```python
# Import necessary libraries
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.linear_model import SGDClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, confusion_matrix, roc_curve, auc, log_loss, silhouette_score
from scipy.stats import mode
from sklearn.preprocessing import label_binarize
```

```python
BASE_PATH = r"C:\Users\asus\OneDrive\Desktop\Machine Project\fruits-360_100x100\fruits-360"
train_dir = os.path.join(BASE_PATH, 'Training')
test_dir  = os.path.join(BASE_PATH, 'Test')
```

```python
classes = ['Apple Red 1', 'Apple Golden 1', 'Cherry 1', 'Orange 1', 'Lemon 1']
```

```python
# Function to load data
def load_data(directory, classes, size=(30, 30)):
    X = []
    y = []
    for label, class_name in enumerate(classes):
        class_path = os.path.join(directory, class_name)
        if not os.path.exists(class_path):
            print(f"Warning: Class directory {class_path} not found.")
            continue
        for img_file in os.listdir(class_path):
            if img_file.lower().endswith(('.jpg', '.jpeg', '.png')):
                img_path = os.path.join(class_path, img_file)
                try:
                    img = Image.open(img_path).convert('L').resize(size)  # Grayscale and resize
                    X.append(np.array(img).flatten())
                    y.append(label)
                except Exception as e:
                    print(f"Error loading {img_path}: {e}")
    return np.array(X), np.array(y)

```

```python
# Load train and test data
train_X, train_y = load_data(train_dir, classes)
test_X, test_y = load_data(test_dir, classes)

# Print dataset information
print(f"Total training samples: {len(train_y)}")
print(f"Total testing samples: {len(test_y)}")
print(f"Class distribution in train: {np.bincount(train_y)}")
print(f"Class distribution in test: {np.bincount(test_y)}")
print(f"Original image size: 100x100 (resized to 30x30 grayscale, features per image: {30*30})")
```

```python
# Logistic Regression (using SGDClassifier for loss curve capability)

# Pipeline for grid search
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA()),
    ('clf', SGDClassifier(loss='log_loss', penalty='l2', max_iter=1000, tol=1e-3, random_state=42))
])

# Grid search for preprocessing (PCA components) and hyperparam (alpha = 1/C)
param_grid = {
    'pca__n_components': [10, 20, 30, 40, 50],
    'clf__alpha': [ 0.01, 0.1, 1.0, 10.0]
}

# Cross-validation: 5-fold on training data
grid_search = GridSearchCV(pipeline, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(train_X, train_y)

print("Best parameters for Logistic:", grid_search.best_params_)

best_n_components_log = grid_search.best_params_['pca__n_components']
best_alpha = grid_search.best_params_['clf__alpha']
```

```python
# Train with loss curve using best params
scaler = StandardScaler().fit(train_X)
train_X_scaled = scaler.transform(train_X)
test_X_scaled = scaler.transform(test_X)

pca_log = PCA(n_components=best_n_components_log).fit(train_X_scaled)
train_X_pca_log = pca_log.transform(train_X_scaled)
test_X_pca_log = pca_log.transform(test_X_scaled)

# SGDClassifier with warm_start for incremental training
clf = SGDClassifier(loss='log_loss', penalty='l2', alpha=best_alpha, max_iter=1, tol=None, warm_start=True, random_state=42)

train_losses = []
test_losses = []
num_epochs = 50

for epoch in range(num_epochs):
    clf.fit(train_X_pca_log, train_y)
    # Train loss
    train_prob = clf.predict_proba(train_X_pca_log)
    train_loss = log_loss(train_y, train_prob)
    train_losses.append(train_loss)
    # Test loss
    test_prob = clf.predict_proba(test_X_pca_log)
    test_loss = log_loss(test_y, test_prob)
    test_losses.append(test_loss)
    
```

```python
# Plot loss curves
plt.figure(figsize=(10, 6))
plt.plot(train_losses, label='Train Loss')
plt.plot(test_losses, label='Test Loss')
plt.xlabel('Epochs')
plt.ylabel('Log Loss')
plt.title('Loss Curves for Logistic Regression')
plt.legend()
plt.show()
```

```python
# Results for Logistic on test data
test_pred_log = clf.predict(test_X_pca_log)
acc_log = accuracy_score(test_y, test_pred_log)
print(f"Test Accuracy for Logistic: {acc_log}")
```

```python
# Confusion Matrix
cm_log = confusion_matrix(test_y, test_pred_log)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_log, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
plt.title('Confusion Matrix for Logistic Regression')
plt.show()
```

```python
# ROC Curve (multi-class One-vs-Rest)
test_y_bin = label_binarize(test_y, classes=range(5))
test_prob_log = clf.predict_proba(test_X_pca_log)

plt.figure(figsize=(10, 8))
for i in range(5):
    fpr, tpr, _ = roc_curve(test_y_bin[:, i], test_prob_log[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f'Class {classes[i]} (AUC = {roc_auc:.2f})')

plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve for Logistic Regression')
plt.legend()
plt.show()
```

```python
# KMeans as Classifier

# Grid search for preprocessing (PCA components) using silhouette score (unsupervised)
scaler = StandardScaler().fit(train_X)  # Reuse scaler
train_X_scaled = scaler.transform(train_X)  # Reuse scaled

n_components_options = [50, 100, 200]
best_sil_score = -1
best_n_components_kmeans = None

for n_comp in n_components_options:
    pca_temp = PCA(n_components=n_comp).fit(train_X_scaled)
    train_X_pca_temp = pca_temp.transform(train_X_scaled)
    kmeans_temp = KMeans(n_clusters=5, random_state=42)
    kmeans_temp.fit(train_X_pca_temp)
    sil_score = silhouette_score(train_X_pca_temp, kmeans_temp.labels_)
    print(f"Silhouette score for n_components={n_comp}: {sil_score}")
    if sil_score > best_sil_score:
        best_sil_score = sil_score
        best_n_components_kmeans = n_comp

print(f"Best n_components for KMeans: {best_n_components_kmeans}")
```

```python
# Train KMeans with best params
pca_kmeans = PCA(n_components=best_n_components_kmeans).fit(train_X_scaled)
train_X_pca_kmeans = pca_kmeans.transform(train_X_scaled)
test_X_pca_kmeans = pca_kmeans.transform(scaler.transform(test_X))

kmeans = KMeans(n_clusters=5, random_state=42)
kmeans.fit(train_X_pca_kmeans)

# Assign clusters to classes based on majority vote
cluster_to_class = np.zeros(5, dtype=int)
for cluster in range(5):
    mask = (kmeans.labels_ == cluster)
    if np.sum(mask) > 0:
        cluster_to_class[cluster] = mode(train_y[mask])[0]

# Predict on test
test_clusters = kmeans.predict(test_X_pca_kmeans)
test_pred_kmeans = cluster_to_class[test_clusters]

acc_kmeans = accuracy_score(test_y, test_pred_kmeans)
print(f"Test Accuracy for KMeans: {acc_kmeans}")

# Confusion Matrix
cm_kmeans = confusion_matrix(test_y, test_pred_kmeans)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_kmeans, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
plt.title('Confusion Matrix for KMeans')
plt.show()

# For KMeans, no probabilistic predictions, so no ROC curve or log loss curve.
# Inertia (loss equivalent): kmeans.inertia_
print(f"KMeans Inertia: {kmeans.inertia_}")
# Note: No loss curve available for KMeans as it's not iterative in the same way; use inertia for different k if needed.
```
